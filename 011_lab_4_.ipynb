```python
print("JOB A")

import random

# Initializations
total_parts = 200 # Simulating 200 parts to observe drift
total_accepted = 0
total_rejected = 0
consecutive_rejects = 0
num_alarms = 0
first_alarm_part = None
shift_value = 0.0 # Initial drift value
alarm_triggered_for_current_streak = False # Flag to ensure alarms are counted once per streak

# Acceptance criteria for diameter (adjusted for sensitivity to 0.05mm drift)
# Original range 24.85 to 25.15. Tightening max_acceptable_dia to 25.10
# to make the 0.05mm drift (shifting range to 24.90-25.20) more likely to cause rejections.
min_acceptable_dia = 24.85
max_acceptable_dia = 25.10 # Adjusted to ensure alarm triggers with 0.05mm drift

print("Starting simulation...")

for part in range(1, total_parts + 1):
    # Apply drift after part 100
    if part > 100:
        shift_value = 0.05 # Shift the random range upward by 0.05 mm

    # Generate diameter with potential drift, rounded to 3 decimal places
    # The base range is (24.85, 25.15) as per previous instructions for the machine's natural variation
    dia_float = random.uniform(24.85 + shift_value, 25.15 + shift_value)
    dia = round(dia_float, 3)

    # Check for acceptance/rejection
    if min_acceptable_dia <= dia <= max_acceptable_dia:
        total_accepted += 1
        consecutive_rejects = 0 # Reset on accept
        alarm_triggered_for_current_streak = False # Reset alarm flag on accept
    else:
        total_rejected += 1
        consecutive_rejects += 1 # Increment on reject

    # Check for tool wear alarm
    # An alarm is raised when consecutive_rejects hits 3 or more for the first time in a streak
    if consecutive_rejects >= 3 and not alarm_triggered_for_current_streak:
        if first_alarm_part is None:
            first_alarm_part = part
        num_alarms += 1
        print(f"⚠ TOOL WEAR SUSPECTED at part {part} (dia: {dia:.3f}). Consecutive rejects: {consecutive_rejects}.")
        alarm_triggered_for_current_streak = True # Set flag to prevent multiple alarms for this specific streak

print("\n--- Simulation Report ---")
print(f"Total parts simulated: {total_parts}")
print(f"Total accepted: {total_accepted}")
print(f"Total rejected: {total_rejected}")

if total_parts > 0:
    acceptance_percentage = (total_accepted / total_parts) * 100
    print(f"Acceptance percentage: {acceptance_percentage:.2f}%")
else:
    print("Acceptance percentage: N/A (no parts simulated)")

print(f"Number of alarms raised: {num_alarms}")
if first_alarm_part is not None:
    print(f"Part number of the first alarm: {first_alarm_part}")
else:
    print("No alarms were raised.")
```
